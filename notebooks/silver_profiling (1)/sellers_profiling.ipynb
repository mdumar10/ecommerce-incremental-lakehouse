```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python
df = spark.table("ecommerce.bronze.sellers")

display(df.limit(10))
df.printSchema()

print("Total rows:", df.count())
```

```sql
%python
print(


    df.select("seller_id").distinct().count()
)
```

```sql
%python
for c in df.columns:
    print(c, df.filter(col(c).isNull()).count())
```

```sql
%python
df.filter(
    (col("seller_zip_code_prefix") < 0) |
    (col("seller_zip_code_prefix") > 99999)
).count()
```
