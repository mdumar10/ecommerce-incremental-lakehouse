```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python
df = spark.table("ecommerce.bronze.geolocation")

display(df.limit(10))
df.printSchema()

print("Total rows:", df.count())
```

```sql
%python
print( df.count())

print(

    df.distinct().count()
)
```

```sql
%python
for c in df.columns:
    print(c, df.filter(col(c).isNull()).count())
```

```sql
%python
df.filter(
    (col("geolocation_lat") < -90) |
    (col("geolocation_lat") > 90)
).count()
```

```sql
%python
df.filter(
    (col("geolocation_lng") < -180) |
    (col("geolocation_lng") > 180)
).count()
```
