```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python
df = spark.table("ecommerce.bronze.order_items")

display(df.limit(10))
df.printSchema()
```

```sql
%python
print("Total rows:", df.count())

print(
    "Distinct order_id + order_item_id:",
    df.select("order_id", "order_item_id").distinct().count()
)
```

```sql
%python

for c in df.columns:
    print(c, df.filter(col(c).isNull()).count())
```

```sql
%python
print("Distinct order_item_id:", df.select("order_item_id").distinct().count())
```

```sql
%python
df.filter(col("price") < 0).count()
```

```sql
%python
df.filter(col("freight_value") < 0).count()
```

```sql
%python
orders = spark.table("ecommerce.bronze.orders")

df.select("order_id").distinct().join(
    orders.select("order_id").distinct(),
    "order_id",
    "left_anti"
).count()
```

```sql
%python
products = spark.table("ecommerce.bronze.products")

df.select("product_id").distinct().join(
    products.select("product_id").distinct(),
    "product_id",
    "left_anti"
).count()
```

```sql
%python
sellers = spark.table("ecommerce.bronze.sellers")

df.select("seller_id").distinct().join(
    sellers.select("seller_id").distinct(),
    "seller_id",
    "left_anti"
).count()
```
