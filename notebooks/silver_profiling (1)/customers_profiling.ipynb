```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python
df = spark.table("ecommerce.bronze.customers")

df.show(10)
```

```sql
%python
df.printSchema()
```

----------------   Grain 1  row → 1 customer_id    ----------------  

```sql
%python

print("Total rows:", df.count())
print("Distinct customer_id:", df.select("customer_id").distinct().count())
print("Distinct customer_unique_id:", df.select("customer_unique_id").distinct().count())
```

- Total rows: 99466

```sql
%python


df.groupBy("customer_id") \
  .count() \
  .filter(col("count") > 1) \
  .show()
```

null check 


```sql
%python


df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).show()
```
