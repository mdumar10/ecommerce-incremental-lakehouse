```sql
SELECT *
FROM ecommerce.silver.customers
LIMIT 10;
```

```sql
SELECT
    customer_unique_id,
    COUNT(*) AS customer_id_count
FROM ecommerce.silver.customers
GROUP BY customer_unique_id
ORDER BY customer_id_count DESC
LIMIT 10;
```

```sql
SELECT
    customer_id,
    customer_unique_id,
    customer_zip_code_prefix,
    customer_city,
    customer_state,
    created_at,
    updated_at
FROM ecommerce.silver.customers
WHERE customer_unique_id = '8d50f5eadf50201ccdcedfb9e2ac8455';
```

```sql
SELECT
    c.customer_unique_id,
    COUNT(DISTINCT c.customer_id) AS customer_id_count,
    COUNT(DISTINCT o.order_id) AS order_count
FROM ecommerce.silver.customers c
LEFT JOIN ecommerce.silver.orders o
    ON c.customer_id = o.customer_id
WHERE c.customer_unique_id = '8d50f5eadf50201ccdcedfb9e2ac8455'
GROUP BY c.customer_unique_id;
```

```sql
SELECT
    COUNT(*) AS unique_customers,
    COUNT(DISTINCT customer_id) AS customer_ids
FROM ecommerce.silver.customers;
```

```sql
SELECT
    customer_state,
    COUNT(*) AS customer_count
FROM ecommerce.silver.customers
GROUP BY customer_state
ORDER BY customer_count DESC;
```

```sql
SELECT
    customer_state,
    COUNT(*) AS customer_count
FROM ecommerce.silver.customers
GROUP BY customer_state
ORDER BY customer_count DESC;
```

```sql
SELECT
    COUNT(DISTINCT customer_zip_code_prefix) AS distinct_zip_codes,
    COUNT(*) AS customer_count
FROM ecommerce.silver.customers;
```

```sql
SELECT
    customer_id,
    COUNT(*) AS order_count
FROM ecommerce.silver.orders
GROUP BY customer_id
ORDER BY order_count DESC
LIMIT 10;
```

```sql
SELECT
    COUNT(*) AS customers_without_orders
FROM ecommerce.silver.customers c
LEFT JOIN ecommerce.silver.orders o
    ON c.customer_id = o.customer_id
WHERE o.customer_id IS NULL;
```
