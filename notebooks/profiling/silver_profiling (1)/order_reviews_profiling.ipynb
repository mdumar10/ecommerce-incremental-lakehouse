```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python

df = spark.table("ecommerce.bronze.order_reviews")

display(df.limit(10))
df.printSchema()
```

```sql
%python
print("Total rows:", df.count())
```

```sql
%python
print(df.select("review_id").distinct().count())

print( df.select("order_id").distinct().count())
```

```sql
%python
print(

    df.select("review_id", "order_id").distinct().count()
)
```

```sql
%python
for c in df.columns:
    print(c, df.filter(col(c).isNull()).count())
```

```sql
%python
df.filter(
    (col("review_score") < 1) |
    (col("review_score") > 5)
).count()
```

```sql
%python
df.filter(
    col("review_answer_timestamp") < col("review_creation_date")
).count()
```
