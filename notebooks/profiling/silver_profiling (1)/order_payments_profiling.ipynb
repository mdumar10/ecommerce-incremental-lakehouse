```sql
%python
from pyspark.sql.functions import col , sum
```

```sql
%python
df = spark.table("ecommerce.bronze.order_payments")

display(df.limit(10))
df.printSchema()
```

```sql
%python
print("Total rows:", df.count())

print(

    df.select("order_id", "payment_sequential").distinct().count()
)
```

```sql
%python
for c in df.columns:
    print(c, df.filter(col(c).isNull()).count())
```

```sql
%python
df.filter(col("payment_value") < 0).count()
```

```sql
%python
df.filter(col("payment_installments") <= 0).count()
```

```sql
%python
df.filter(col("payment_value") < 0).count()
```

```sql
%python
orders = spark.table("ecommerce.bronze.orders")

df.select("order_id").distinct().join(
    orders.select("order_id").distinct(),
    "order_id",
    "left_anti"
).count()
```
