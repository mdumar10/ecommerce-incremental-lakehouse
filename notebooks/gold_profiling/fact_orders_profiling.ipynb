Fact Orders — Gold Profiling

Goal

Build a Gold table with one row per order.

Source tables

ecommerce.silver.orders

ecommerce.silver.order_items

ecommerce.silver.order_payments

ecommerce.silver.order_reviews

ecommerce.silver.customers

```sql
SELECT *
FROM ecommerce.silver.orders
LIMIT 10;
```

order_items


```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT order_id) AS distinct_orders
FROM ecommerce.silver.orders;
```

```sql
SELECT
    order_id,
    COUNT(*) AS item_rows
FROM ecommerce.silver.order_items
GROUP BY order_id
ORDER BY item_rows DESC;
```

```sql
SELECT
    order_id,
    COUNT(*) AS item_count,
    SUM(price) AS total_product_value,
    SUM(freight_value) AS total_freight_value
FROM ecommerce.silver.order_items
GROUP BY order_id
ORDER BY item_count DESC
LIMIT 10;
```

```sql
SELECT
    order_item_id,
    product_id,
    price,
    freight_value
FROM ecommerce.silver.order_items
WHERE order_id = '8272b63d03f5f79c56e9e4120aec44ef'
ORDER BY order_item_id;
```

order_payments


```sql
SELECT
    order_id,
    COUNT(*) AS payment_count,
    SUM(payment_value) AS total_payment_value
FROM ecommerce.silver.order_payments
GROUP BY order_id
ORDER BY payment_count DESC
LIMIT 10;
```

```sql
SELECT
    payment_sequential,
    payment_type,
    payment_installments,
    payment_value
FROM ecommerce.silver.order_payments
WHERE order_id = 'fa65dad1b0e818e3ccc5cb0e39231352'
ORDER BY payment_sequential;
```

order_reviews

```sql
SELECT
    order_id,
    COUNT(*) AS review_count,
    AVG(review_score) AS average_review_score
FROM ecommerce.silver.order_reviews
GROUP BY order_id
ORDER BY review_count DESC
LIMIT 10;
```

```sql
SELECT
    COUNT(*) AS rows_after_join,
    COUNT(DISTINCT o.order_id) AS distinct_orders
FROM ecommerce.silver.orders o
LEFT JOIN ecommerce.silver.order_items i
    ON o.order_id = i.order_id
LEFT JOIN ecommerce.silver.order_payments p
    ON o.order_id = p.order_id
LEFT JOIN ecommerce.silver.order_reviews r
    ON o.order_id = r.order_id;
```
