```sql
SELECT
    COUNT(*) AS total_rows,
    COUNT(DISTINCT product_id) AS distinct_products
FROM ecommerce.silver.products;
```

```sql
SELECT *
FROM ecommerce.silver.products
LIMIT 5;
```

```sql
SELECT
    COUNT(DISTINCT product_id) AS products_in_order_items,
    COUNT(*) AS total_order_item_rows
FROM ecommerce.silver.order_items;
```

```sql
SELECT DISTINCT
    i.product_id
FROM ecommerce.silver.order_items i
LEFT JOIN ecommerce.silver.products p
    ON i.product_id = p.product_id
WHERE p.product_id IS NULL;
```

```sql
SELECT *
FROM ecommerce.silver.order_items
LIMIT 5;
```

```sql
SELECT
    product_id,
    order_id,
    COUNT(*) AS item_count
FROM ecommerce.silver.order_items
GROUP BY product_id, order_id
HAVING COUNT(*) > 1
LIMIT 10;
```
